```python
# Smart Home Agent 

# Take sensor inputs
temperature = float(input("Enter temperature (°C): "))
light_intensity = int(input("Enter light intensity (0-100): "))
time_of_day = input("Enter time of day (morning/afternoon/evening/night): ").lower()
humidity = int(input("Enter humidity (%): "))

 #  Set default device status
fan = "OFF"
ac = "OFF"
light = "OFF"
heater = "OFF"

 #  Fan
if temperature > 30 and humidity < 70:
    fan = "ON"

 #  AC (used instead of fan when it's hot AND humid)
if temperature > 32 and humidity >= 70:
    ac = "ON"
    fan = "OFF"

 #  Light
if time_of_day in ("evening", "night") and light_intensity < 40:
    light = "ON"

 #  Heater
if time_of_day == "night" and temperature < 15:
    heater = "ON"
 
 #  Show the agent's decisions
print("\n--- Smart Home Agent Decision ---")
print("Fan            :", fan)
print("AC             :", ac)
print("Light          :", light)
print("Heater         :", heater)
 
```

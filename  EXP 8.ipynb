```python
# --------------------------------------------------------
# Rule-Based Chatbot using NLTK
# --------------------------------------------------------

# Install NLTK (Run once)
# pip install nltk

import nltk
from nltk.chat.util import Chat, reflections

# Define chatbot patterns and responses
pairs = [
    [
        r"hi|hello|hey",
        ["Hello! How can I help you today?",
         "Hi there! What can I do for you?"]
    ],
    [
        r"what is your name ?",
        ["I am an AI Chatbot built using NLTK."]
    ],
    [
        r"how are you ?",
        ["I am doing well. Thank you for asking!"]
    ],
    [
        r"what can you do ?",
        ["I can answer simple questions and chat with you."]
    ],
    [
        r"who developed you ?",
        ["I was developed using Python and the NLTK library."]
    ],
    [
        r"what is python ?",
        ["Python is a high-level, interpreted programming language."]
    ],
    [
        r"what is ai ?",
        ["Artificial Intelligence (AI) enables machines to simulate human intelligence."]
    ],
    [
        r"bye|exit|quit",
        ["Goodbye! Have a great day.",
         "Bye! Take care."]
    ],
    [
        r"(.*)",
        ["Sorry, I don't understand that.",
         "Can you please rephrase your question?"]
    ]
]

# Create chatbot
chatbot = Chat(pairs, reflections)

# Welcome message
print("======================================")
print("        AI Rule-Based Chatbot")
print("======================================")
print("Type 'bye' or 'exit' to quit.\n")

# Start conversation
if __name__ == "__main__":
    chatbot.converse()
```

```python
# ----------------------------------------------------
# Basic Fuzzy Inference System for Fan Speed Control
# ----------------------------------------------------

# Membership functions
def low_temp(temp):
    if temp <= 20:
        return 1.0
    elif 20 < temp < 30:
        return (30 - temp) / 10
    else:
        return 0.0

def medium_temp(temp):
    if 20 < temp < 30:
        return (temp - 20) / 10
    elif 30 <= temp <= 40:
        return (40 - temp) / 10
    else:
        return 0.0

def high_temp(temp):
    if temp <= 30:
        return 0.0
    elif 30 < temp < 40:
        return (temp - 30) / 10
    else:
        return 1.0

# Main Program
temperature = float(input("Enter Temperature (°C): "))

# Fuzzification
low = low_temp(temperature)
medium = medium_temp(temperature)
high = high_temp(temperature)

print("\nMembership Values")
print("----------------------------")
print(f"Low    : {low:.2f}")
print(f"Medium : {medium:.2f}")
print(f"High   : {high:.2f}")

# Inference Engine
if low >= medium and low >= high:
    fan_speed = "LOW"
elif medium >= low and medium >= high:
    fan_speed = "MEDIUM"
else:
    fan_speed = "HIGH"
    
print(f"\nCalculated Fan Speed: {fan_speed}")
```

```python
# ---------------------------------------------------------
# Government Scheme Eligibility Checker
# ---------------------------------------------------------

# List of eligible regions
eligible_regions = [
    "Maharashtra",
    "Karnataka",
    "Gujarat",
    "Goa"
]

# List of acceptable educational qualifications
education_levels = [
    "High School",
    "Diploma",
    "Graduate",
    "Post Graduate",
    "PhD"
]

# Income limit
INCOME_LIMIT = 300000


# -------------------- User Input --------------------

name = input("Enter Applicant Name: ")

age = int(input("Enter Age: "))

income = float(input("Enter Annual Income (Rs.): "))

region = input("Enter Region/State: ")

education = input("Enter Highest Qualification: ")


# ---------------- Eligibility Check -----------------

age_ok = age >= 18
income_ok = income <= INCOME_LIMIT
region_ok = region in eligible_regions
education_ok = education in education_levels


# -------------------- Decision ----------------------

if age_ok and income_ok and region_ok and education_ok:
    print("\nResult: ELIGIBLE")
    print("Congratulations! You are eligible for the government scheme.")

elif age_ok and education_ok and (income_ok or region_ok):
    print("\nResult: CONDITIONALLY ELIGIBLE")
    print("Your application requires further verification.")

else:
    print("\nResult: NOT ELIGIBLE")
    print("You do not satisfy the eligibility criteria.")


# ---------------- Applicant Summary -----------------

print("\n----------- Applicant Details -----------")
print("Name :", name)
print("Age :", age)
print("Income : Rs.", income)
print("Region :", region)
print("Education :", education)
```

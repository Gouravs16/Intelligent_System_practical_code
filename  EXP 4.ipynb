```python
# -----------------------------------------------------
# Expert System for Medical Diagnosis
# Knowledge Base + Inference Engine
# -----------------------------------------------------

# Knowledge Base (Disease : Symptoms)
knowledge_base = {
    "Common Cold": ["cough", "sneezing", "sore throat"],
    "Flu": ["fever", "cough", "body pain", "headache"],
    "COVID-19": ["fever", "cough", "loss of taste", "breathing difficulty"],
    "Malaria": ["fever", "chills", "sweating", "headache"],
    "Dengue": ["fever", "joint pain", "rash", "headache"],
    "Typhoid": ["fever", "abdominal pain", "weakness", "loss of appetite"]
}


# Inference Engine
def diagnose(user_symptoms):
    possible_diseases = {}

    for disease, symptoms in knowledge_base.items():
        match_count = 0

        for symptom in symptoms:
            if symptom in user_symptoms:
                match_count += 1

        possible_diseases[disease] = match_count

    # Find disease with maximum symptom match
    diagnosis = max(
        possible_diseases,
        key=possible_diseases.get
    )

    score = possible_diseases[diagnosis]

    if score == 0:
        return None
    else:
        return diagnosis, score


# Main Program
print("===================================")
print("     MEDICAL DIAGNOSIS EXPERT SYSTEM")
print("===================================")

print("\nAvailable Symptoms:")

all_symptoms = set()

for symptoms in knowledge_base.values():
    all_symptoms.update(symptoms)

for symptom in sorted(all_symptoms):
    print("-", symptom)


print("\nEnter symptoms separated by commas")

user_input = input("Symptoms: ")

user_symptoms = [
    s.strip().lower()
    for s in user_input.split(",")
]


result = diagnose(user_symptoms)


print("\n---------- Diagnosis ----------")

if result is None:
    print("No matching disease found.")
else:
    disease, score = result
    print("Possible Disease :", disease)
    print("Matching Symptoms:", score)


print("--------------------------------")
print("Note: This is an educational expert system and")
print("should not replace professional medical advice.")
```
